```python
def segment_customer(row):
    r = int(row['R_Score'])
    f = int(row['F_Score'])
    m = int(row['M_Score'])
    if r >= 4 and f >= 4 and m >= 4:
        return 'Best'
    elif r <= 2 and f >= 3:
        return 'At Risk'
    elif r <= 2 and f <= 2:
        return 'Lost'
    elif r >= 4 and f <= 2:
        return 'Almost loyal'
    elif f >= 4:
        return 'Loyal '
    else:
        return 'Others'

rfm['Segment'] = rfm.apply(segment_customer, axis=1)
rfm[['R_Score', 'F_Score', 'M_Score', 'Segment']].head(10)
```

```python
%%writefile customer_analysis.py
import pandas as pd
import numpy as np
def create_rfm(df):
    snapshot_date = df['InvoiceDate'].max() + pd.Timedelta(days=1)
    rfm = df.groupby('CustomerID').agg({
        'InvoiceDate': lambda x: (snapshot_date - x.max()).days,
        'InvoiceNo': 'nunique',
        'Revenue': 'sum'
    })
    rfm.columns = ['Recency', 'Frequency', 'Monetary']
    return rfm
```

```python
## Project Findings

1. The analysis identified 5,878 unique customers from the cleaned transaction dataset.

2. The repeat-purchase rate was 72.39%, indicating that a substantial proportion of customers made multiple purchases.

3. The Best customer segment generated the highest revenue of £11,819,626.88, with an average revenue of £9,169.61 per customer.

4. The Lost segment contained the largest number of customers (1,526), but had the lowest average revenue of £429.52.

5. High-spending outliers accounted for 65.29% of total revenue, showing that revenue is concentrated among a relatively small group of customers.

6. Including negative-quantity transactions reduced revenue by £1,084,812.98, approximately 6.25%.

7. Average revenue per customer increased from £1,343.20 over six months to £2,955.90 over the full observation period.

8. Excluding high-spending outliers reduced average customer revenue from £2,955.90 to £1,149.68.

## Business Recommendations

- Focus on retaining Best and Loyal customers through loyalty programmes and personalised offers.
- Develop re-engagement campaigns for Lost customers to encourage repeat purchases.
- Target At Risk customers with special offers before they become inactive.
- Monitor high-value customers carefully because they contribute a significant share of total revenue.
- Track product returns to understand their impact on overall revenue.
- Use monthly cohort retention analysis to evaluate whether newly acquired customers continue purchasing over time.
```

```python
## Limitations and Assumptions

1. The analysis is based on historical transaction data and does not guarantee future customer behaviour.

2. Customer Lifetime Value (CLV) is estimated using historical revenue per customer over the available observation period. It does not include profit margins, acquisition costs or future revenue forecasts.

3. Transactions without Customer IDs were excluded because they could not be linked to individual customers.

4. Cancelled invoices and invalid transactions were excluded from the baseline revenue analysis. Negative-quantity transactions were included separately for returns sensitivity analysis.

5. High-spending customers were identified using the Interquartile Range (IQR) method. These customers are not necessarily errors or fraudulent transactions.

6. Different observation windows contain different customer populations, which can affect average revenue comparisons.

7. The analysis does not include marketing expenses, customer acquisition costs or customer demographics.

8. The findings are limited to the available Online Retail II dataset and may not represent every retail business.
```
